```python
import cv2
```

```python
cap = cv2.VideoCapture(0)
```

```python
cap.isOpened()
```

```python
cap.set(3,1280)
cap.set(4,720)
```

```python
print(cap.get(3))
print(cap.get(4))
```

```python

while True:
    # Read frame-by-frame
    ret, frame = cap.read()
    
    if not ret:
        break

    # Display the frame
    cv2.imshow('Webcam Feed', frame)

    # Exit on pressing 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Cleanup
cap.release()
cv2.destroyAllWindows()
```
